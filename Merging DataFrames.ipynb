```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import pickle

%load_ext autoreload
%autoreload 2

% matplotlib inline

plt.rcParams['figure.figsize'] = (10, 8)
```

```python
with open('daily_BSH32.pickle', 'rb') as f:
    BSH_dict = pickle.load(f)
    
with open('daily_rad32.pickle', 'rb') as f:
    RAD_dict = pickle.load(f)

with open('daily_TEMP32.pickle', 'rb') as f:
    TMP_dict = pickle.load(f)
    
assert len(BSH_dict) == len(RAD_dict)
assert len(BSH_dict) == len(TMP_dict)
assert len(TMP_dict) == len(RAD_dict)
```

```python
print(RAD_dict['Van Bölge'].head())
print(RAD_dict['Tarsus'].head())
print(BSH_dict['Van Bölge'].head())
print(BSH_dict['Tarsus'].head())
print(TMP_dict['Van Bölge'].head())
print(TMP_dict['Tarsus'].head())
```

```python
aks_rad = RAD_dict['Aksaray']
aks_bsh = BSH_dict['Aksaray']
aks_tmp = TMP_dict['Aksaray']

print(len(aks_rad))
print(len(aks_bsh))
print(len(aks_tmp))
```

```python
merged = pd.concat([aks_rad, aks_bsh, aks_tmp], axis=1)
print(merged.head())
print(len(merged))
print(len(merged.dropna(how='any')))
```

```python
merged = merged.loc[:, ~merged.columns.duplicated()]
print(len(merged))
print(merged.head())
print(merged.tail())
```

```python
keys = RAD_dict.keys()
merged_dict = {}
for key in keys:
    rad_entries = len(RAD_dict[key])
    bsh_entries = len(BSH_dict[key])
    tmp_entries = len(TMP_dict[key])
    print('Station {} has {} RAD entries, {} BSH entries, {} TMP entries'.format(
        key, rad_entries, bsh_entries, tmp_entries))
    merger = pd.concat([BSH_dict[key], RAD_dict[key], TMP_dict[key]], axis=1)
    merger = merger.loc[:, ~merger.columns.duplicated()]
    print('Station {} has {} merged entries in total'.format(key, len(merger)))
    merger.dropna(how='any', inplace=True)
    print('Station {} has {} merged non-NA entries in total'.format(key, len(merger)))
    
    merged_dict.update({key:merger})
```

```python
print(merged_dict['Artvin'].head())
```

```python
filename = 'daily32_RAD_BSH_TMP.pickle'
with open(filename, 'wb') as f:
    pickle.dump(merged_dict, f)
```
