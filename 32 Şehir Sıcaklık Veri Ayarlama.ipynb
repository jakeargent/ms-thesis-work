```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import pickle
import os

%load_ext autoreload
%autoreload 2

% matplotlib inline

plt.rcParams['figure.figsize'] = (10, 8)
```

```python
filepath = 'saatlik_sicaklik_(T-c).txt'
temperature = pd.read_csv(filepath, sep=';')
print(temperature.head())
```

```python
temperature.rename(index=str, columns={"Yil":"Year", "Ay":"Month", "Gün":"Day", "Saat (UTC)":"Hour"},
                  inplace=True)
print(temperature.head())
```

```python
print(pd.to_datetime(temperature[['Year', 'Month', 'Day', 'Hour']], format='%Y-%M-%D-%H')[:10])
```

```python
datas = {'Station':temperature['Istasyon No'],
         'Date':pd.to_datetime(temperature[['Year', 'Month', 'Day']], format='%Y-%M-%D'),
         'Temperature(avg)':temperature['Sıcaklık (°C)']}
temperature_proper = pd.DataFrame(datas)
print(temperature_proper[:25])
```

```python
temperature_proper = temperature_proper.reset_index().groupby(['Station', 'Date']).mean()
print(temperature_proper.head())
```

```python
station_no_to_name = {17040:'Rize', 17045:'Artvin', 17052:'Kırklareli', 17070:'Bolu', 17074:'Kastamonu',
                      17086:'Tokat', 17088:'Gümüşhane', 17097:'Kars', 17099:'Ağrı', 17112:'Çanakkale',
                      17116:'Bursa', 17162:'Gemerek', 17172:'Van Bölge', 17190:'Afyonkarahisar Bölge',
                      17192:'Aksaray', 17199:'Malatya', 17239:'Akşehir', 17240:'Isparta', 17242:'Beyşehir',
                      17246:'Karaman', 17262:'Kilis', 17285:'Hakkari', 17287:'Şırnak', 17351:'Adana Bölge',
                      17624:'Ünye', 17684:'Suşehri', 17688:'Tortum', 17718:'Tercan', 17720:'Doğubeyazıt',
                      17722:'Burhaniye', 17734:'Divriği', 17749:'Kemalpaşa', 17754:'Kulu', 17760:'Boğazlıyan',
                      17776:'Solhan', 17780:'Malazgirt', 17789:'Menemen', 17806:'Palu', 17836:'Develi',
                      17847:'Ergani', 17866:'Göksun', 17906:'Ulukışla', 17944:'Bozova', 17952:'Elmalı',
                      17968:'Ceylanpınar Tigem', 17978:'Tarsus'}
print(len(station_no_to_name.keys()))
```

```python
temperature_proper = temperature_proper.reset_index()
print(temperature_proper.head())
```

```python
print(temperature_proper.loc[temperature_proper['Station']==17040][:10])
```

```python
trial = temperature_proper.loc[temperature_proper['Station']==17040][:10]
trial.loc[trial['Station']==17040, 'Station'] = 'Rize'
print(trial)
```

```python
for key in station_no_to_name:
    temperature_proper.loc[temperature_proper['Station']==key, 'Station'] = station_no_to_name[key]

print(temperature_proper.head())
print(temperature_proper.tail())
```

```python
try_df = temperature_proper.loc[temperature_proper['Station']=='Artvin'].copy()
try_df.set_index('Date', inplace=True)
print(try_df.head())
```

```python
temperature_dict = {}
for key in station_no_to_name:
    # pick the subset of data where Station == <Station> and assign it to a new DataFrame
    # Then set the index on it to 'Date', and add it to the dictionary
    new_df = temperature_proper.loc[temperature_proper['Station']==station_no_to_name[key]].copy()
    new_df.set_index('Date', inplace=True)
    temperature_dict.update({station_no_to_name[key]:new_df})

print(len(temperature_dict))
```

```python
artvin = temperature_dict['Artvin']
print(artvin.head())
print(artvin.tail())
aksaray = temperature_dict['Aksaray']
print(aksaray.head())
print(aksaray.tail())
```

```python
filename = 'daily_TEMP32.pickle'
with open(filename, 'wb') as f:
    pickle.dump(temperature_dict, f)
```
