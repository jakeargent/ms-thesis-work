```python
import pickle
import numpy as np
import pandas as pd

%load_ext autoreload
%autoreload 2
```

```python
def print_stuff(filespec):
    fname = 'measures_6to31_' + filespec + '.pkl'
    with open(fname, 'rb') as f:
        measures = pickle.load(f)
        
    mess = ['rmse', 'mae', 'r2', 'evs', 'fie_h', 'fie_o', 'fie_oh', 'lin']
    test6 = np.array(measures['te6'])
    test31 = np.array(measures['te31'])
    
    assert test6.shape == test31.shape
    assert test6.shape == (50,8)
    
    print(filespec.upper())
    print('Mean performances')
    means6 = test6.mean(axis=0, keepdims=True)
    means31 = test31.mean(axis=0, keepdims=True)
    means = np.concatenate([means6, means31])
    df_mean = pd.DataFrame(means, columns=mess, index=['6 Stat', '31 Stat'])
    print(df_mean)
    
    max6 = test6.max(axis=0, keepdims=True)
    max31 = test31.max(axis=0, keepdims=True)
    maxes = np.concatenate([max6, max31])
    
    min6 = test6.min(axis=0, keepdims=True)
    min31 = test31.min(axis=0, keepdims=True)
    mins = np.concatenate([min6, min31])
    
    best = np.empty((2,8))
    best[:, 0:1] = mins[:, 0:1]
    best[:, 2:] = maxes[:, 2:]
    df_best = pd.DataFrame(best, columns=mess, index=['6 Stat', '31 Stat'])
    print('\nBest Performances:')
    print(df_best)
    print('_'*50 + '\n')
```

```python
print_stuff('nN_hh0')
print_stuff('nN_h')
print_stuff('nNPhi_hh0')
print_stuff('nNPhi_h')
print_stuff('nNT_hh0')
print_stuff('nNT_h')
print_stuff('nNTPhi_hh0')
print_stuff('nNTPhi_h')
```

```python
def ann_subset_perf(filespec):
    fname = 'measures_6to31_' + filespec + '.pkl'
    with open(fname, 'rb') as f:
        measures = pickle.load(f)
        
    mess = ['RMSE', 'MAE', 'R2', 'EVS', 'FIE_H', 'FIE_O', 'FIE_OH', 'LINCC']
    
    varset, target = filespec.split('_')
    if target == 'hh0':
        indices = [varset + ' 6 Stat', varset + ' 31 Stat']
    elif target == 'h':
        indices = [varset + ' H 6 Stat', varset + ' H 31 Stat']
    
    
    test6 = np.array(measures['te6'])
    test31 = np.array(measures['te31'])
    
    assert test6.shape == test31.shape
    assert test6.shape == (50,8)
    
    #print(filespec.upper())
    #print('Mean performances')
    means6 = test6.mean(axis=0, keepdims=True)
    means31 = test31.mean(axis=0, keepdims=True)
    stds6 = test6.std(axis=0, keepdims=True)
    stds31 = test31.std(axis=0, keepdims=True)
    means = np.concatenate([means6, means31])
    stds = np.concatenate([stds6, stds31])
    df_mean = pd.DataFrame(means, columns=mess, index=indices)
    df_std = pd.DataFrame(stds, columns=mess, index=indices)
    return df_mean
```

```python
tries = ['nN_hh0', 'nN_h', 'nNPhi_hh0', 'nNPhi_h', 'nNT_hh0', 'nNT_h', 'nNTPhi_hh0', 'nNTPhi_h']

for trial in tries:
    if trial == 'nN_hh0':
        table = ann_subset_perf(trial)
    else:
        table = pd.concat([table, ann_subset_perf(trial)])
    
print(table)
```
